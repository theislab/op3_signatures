```python
import pubchempy as pcp
from tqdm import tqdm
```

```python
df_dili = pd.read_pickle('../../dili.pkl')
```

```python
df_dili = df_dili.drop(columns=['cmap_name', 'symbol', 'code', 'symbol_', 'LPM_emb'])
```

```python
df_dili['pubchem_cid_key'] = df_dili['pubchem_cid'].copy().astype(object)
```

```python
df_dili.loc[~df_dili['pubchem_cid_key'].isna(), 'pubchem_cid_key'] = df_dili.loc[~df_dili['pubchem_cid_key'].isna(), 'pubchem_cid_key'].astype(int).astype(str)
```

```python
print(1)
```

```python
import os
os.makedirs('./dili_single_run_all_l1000_25_50_epochs/', exist_ok=True)
epoch_dirs = os.listdir('../../../lpm_style/files/single_run_all_50_epochs/')
epoch_dirs.remove('last')

for epoch_dir in epoch_dirs:
    df_pert_all = pd.read_pickle(f'../../../lpm_style/files/single_run_all_50_epochs/{epoch_dir}/df_pert.pkl')\
                    .rename(columns={'symbol': 'symbol_all', 
                                     'code': 'code_all', 
                                     'lpm_style_embeddings': 
                                     'lpm_style_embeddings_all'})

    df_pert_l1000 = pd.read_pickle(f'../../../lpm_style/files/single_run_l1000_50_epochs/{epoch_dir}/df_pert.pkl')\
                    .rename(columns={'symbol': 'symbol_l1000', 
                                     'code': 'code_l1000', 
                                     'lpm_style_embeddings': 
                                     'lpm_style_embeddings_l1000'})

    
    df_dili_merged = df_dili.merge(df_pert_all, left_on='pubchem_cid_key', right_on='symbol_all', how='left')\
                            .merge(df_pert_l1000, left_on='pubchem_cid_key', right_on='symbol_l1000', how='left')\
                            .drop(columns=['pubchem_cid_key'])
    
    
    tag = int(epoch_dir.split('_')[-1])
    df_dili_merged.to_pickle(f"./dili_single_run_all_l1000_25_50_epochs/dili_lpm_style_embeddings_epoch_{tag + 1}.pkl")
```
