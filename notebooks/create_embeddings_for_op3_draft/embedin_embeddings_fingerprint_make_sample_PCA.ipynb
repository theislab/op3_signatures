```python
import anndata as ad
import pandas as pd
import numpy as np
```

```python
import os
```

```python
de_train = ad.read_h5ad('./data/benchmark/resources/datasets/neurips-2023-data/de_train.h5ad')
de_test = ad.read_h5ad('./data/benchmark/resources/datasets/neurips-2023-data/de_test.h5ad')
id_map = pd.read_csv('./data/benchmark/resources/datasets/neurips-2023-data/id_map.csv')
```

```python
df_emb = pd.read_pickle('./notebooks/pca_embeddings/op3_PCA_emb.pkl')
```

```python
df_emb_i[~df_emb_i['matched_l1000_idx'].isna()]
```

```python
emb_combinations = [('v1', 64),
 ('v2', 64),
 ('v3', 64),
 ('v4', 64),
 ('v2', 128),
 ('v3', 128),
 ('v4', 128),
]

for item in emb_combinations:
    path = "./data/benchmark/resources/datasets/neurips-2023-data-subsample-pca/" + item[0] + '_' + str(item[1])
    os.makedirs(path, exist_ok=True)
    df_emb_i = df_emb[(df_emb['version'] == item[0])&(df_emb['dim']==item[1])].drop_duplicates(['perturbagen'])
    
    de_train[de_train.obs['sm_name'].isin(df_emb_i[~df_emb_i['matched_l1000_idx'].isna()]['perturbagen'])].write_h5ad(path + '/de_train.h5ad', compression='gzip')
    de_test[de_test.obs['sm_name'].isin(df_emb_i[~df_emb_i['matched_l1000_idx'].isna()]['perturbagen'])].write_h5ad(path + '/de_test.h5ad', compression='gzip')
    id_map[id_map['sm_name'].isin(df_emb_i[~df_emb_i['matched_l1000_idx'].isna()]['perturbagen'])].to_csv(path + '/id_map.csv', index=False)

    df_emb_i[~df_emb_i['matched_l1000_idx'].isna()].reset_index(drop=True).to_pickle(path + "/op3_emb.pkl")
    
```
